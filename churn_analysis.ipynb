```python
print(df['Phone Service'].value_counts())
print(df[df['Phone Service']=='No']['Multiple Lines'].isnull().sum())
print()
print(df['Internet Service'].value_counts())
print(df[df['Internet Service']=='No']['Internet Type'].isnull().sum())
```

```python
data_dict = pd.read_csv('telecom_data_dictionary.csv', encoding='latin1')
data_dict
```

```python
df['Avg Monthly Long Distance Charges'] = df['Avg Monthly Long Distance Charges'].fillna(0)
df['Multiple Lines'] = df['Multiple Lines'].fillna('No Phone Service')
```

```python
df['Internet Type'] = df['Internet Type'].fillna('No Internet Service')

internet_yesno_cols = ['Online Security', 'Online Backup', 'Device Protection Plan',
                        'Premium Tech Support', 'Streaming TV', 'Streaming Movies',
                        'Streaming Music', 'Unlimited Data']

for col in internet_yesno_cols:
    df[col] = df[col].fillna('No Internet Service')

df['Avg Monthly GB Download'] = df['Avg Monthly GB Download'].fillna(0)
```

```python
df['Offer'] = df['Offer'].fillna('No Offer')
df['Churn Category'] = df['Churn Category'].fillna('N/A - Stayed')
df['Churn Reason'] = df['Churn Reason'].fillna('N/A - Stayed')
```

```python
df.isnull().sum().sum()
```

```python
df.isnull().sum().sum()
```

```python
df['Churn_Flag'] = df['Customer Status'].apply(lambda x: 1 if x == 'Churned' else 0)
```

```python
df['Churn_Flag'].value_counts()
print('Churn Rate:', df['Churn_Flag'].mean())
```

```python
df['Tenure_Bucket'] = pd.cut(df['Tenure in Months'],
                              bins=[0, 12, 24, 48, 72],
                              labels=['New', 'Developing', 'Established', 'Loyal'])
```

```python
print(df['Tenure_Bucket'].value_counts())
print()
print(df.groupby('Tenure_Bucket', observed=True)['Churn_Flag'].mean())
```

```python
df['Age_Bucket'] = pd.cut(df['Age'],
                           bins=[18, 30, 50, 65, 80],
                           labels=['Young Adult', 'Mid Career', 'Pre Senior', 'Senior'])
```

```python
print(df['Age_Bucket'].value_counts())
print()
print(df.groupby('Age_Bucket', observed=True)['Churn_Flag'].mean())
```

```python
print('Churn rate, age 61-80:', df[(df['Age'] >= 61) & (df['Age'] <= 80)]['Churn_Flag'].mean())
print('Churn rate, age 66-80:', df[(df['Age'] >= 66) & (df['Age'] <= 80)]['Churn_Flag'].mean())
```

```python
df['Age_Bucket'] = pd.cut(df['Age'],
                           bins=[18, 35, 50, 60, 80],
                           labels=['Young Adult', 'Mid Career', 'Pre Senior', 'Senior'])
```

```python
print(df['Age_Bucket'].value_counts())
print()
print(df.groupby('Age_Bucket', observed=True)['Churn_Flag'].mean())
```

```python
df['Charge_Tier'] = pd.cut(df['Monthly Charge'],
                            bins=[0, 45, 75, 200],
                            labels=['Budget', 'Mid', 'Premium'])
```

```python
print(df['Charge_Tier'].value_counts())
print()
print(df.groupby('Charge_Tier', observed=True)['Churn_Flag'].mean())
print()
print(df.groupby('Charge_Tier', observed=True)['Churn_Flag'].sum())
```

```python
df['Annual_Revenue_at_Risk'] = df.apply(
    lambda row: row['Monthly Charge'] * 12 if row['Churn_Flag'] == 1 else 0,
    axis=1
)
```

```python
print('Total Annual Revenue at Risk: $', round(df['Annual_Revenue_at_Risk'].sum(), 2))
```

```python
df['Churn Category'].value_counts()
```

```python
def categorize_churn(category):
    if category == 'Competitor':
        return 'Competitive Threat'
    elif category == 'N/A - Stayed':
        return 'N/A - Stayed'
    else:
        return 'Fixable'

df['Churn_Type'] = df['Churn Category'].apply(categorize_churn)
```

```python
print(df['Churn_Type'].value_counts())
print()
print(df.groupby('Churn_Type', observed=True)['Annual_Revenue_at_Risk'].sum())
```

```python
df.to_csv('cleaned_telecom_churn.csv', index=False)
print('Saved. Shape:', df.shape)
```

```python
top_reasons = df[df['Customer Status'] == 'Churned']['Churn Reason'].value_counts().head(10)
top_reasons_df = top_reasons.reset_index()
top_reasons_df.columns = ['Churn_Reason', 'Count']
top_reasons_df['Churn_Type'] = top_reasons_df['Churn_Reason'].apply(
    lambda x: 'Competitive Threat' if 'Competitor' in x else 'Fixable'
)
top_reasons_df
```

```python
color_map2 = {'Fixable': '#f39c12', 'Competitive Threat': '#e74c3c'}
bar_colors = [color_map2[t] for t in top_reasons_df['Churn_Type']]

plt.figure(figsize=(9, 6))
plt.barh(top_reasons_df['Churn_Reason'], top_reasons_df['Count'], color=bar_colors)
plt.gca().invert_yaxis()
plt.title('Top 10 Churn Reasons: Fixable (Orange) vs Competitive (Red)', fontsize=13, fontweight='bold')
plt.xlabel('Number of Customers')
plt.show()
```

```python
contract_stats = df.groupby('Contract').agg(
    Churn_Rate=('Churn_Flag', 'mean'),
    Customers=('Churn_Flag', 'count'),
    Revenue_at_Risk=('Annual_Revenue_at_Risk', 'sum')
).reset_index()
contract_stats
```

```python
fig, ax1 = plt.subplots(figsize=(8, 5))

ax1.bar(contract_stats['Contract'], contract_stats['Churn_Rate'], color='#3498db', alpha=0.8)
ax1.set_ylabel('Churn Rate', color='#3498db')
ax1.set_title('Contract Type: Churn Rate vs Revenue at Risk', fontsize=13, fontweight='bold')

ax2 = ax1.twinx()
ax2.plot(contract_stats['Contract'], contract_stats['Revenue_at_Risk'], color='#e74c3c', marker='o', linewidth=2)
ax2.set_ylabel('Annual Revenue at Risk ($)', color='#e74c3c')

plt.show()
```

```python
tenure_order = ['New', 'Developing', 'Established', 'Loyal']
tenure_churn = df.groupby('Tenure_Bucket', observed=True)['Churn_Flag'].mean().reindex(tenure_order)

plt.figure(figsize=(7, 5))
plt.plot(tenure_churn.index, tenure_churn.values, marker='o', linewidth=2.5, color='#8e44ad')
for i, v in enumerate(tenure_churn.values):
    plt.text(i, v + 0.01, f'{v:.1%}', ha='center', fontweight='bold')
plt.title('Churn Risk Drops Sharply as Tenure Increases', fontsize=13, fontweight='bold')
plt.ylabel('Churn Rate')
plt.ylim(0, 0.7)
plt.show()
```

```python
offer_stats = df.groupby('Offer', observed=True).agg(
    Churn_Rate=('Churn_Flag', 'mean'),
    Customers=('Churn_Flag', 'count')
).reset_index().sort_values('Churn_Rate')

colors_offer = ['#e74c3c' if o == 'Offer E' else '#3498db' for o in offer_stats['Offer']]

plt.figure(figsize=(8, 5))
plt.bar(offer_stats['Offer'], offer_stats['Churn_Rate'], color=colors_offer)
plt.title('Offer Performance: Offer E Is Actively Harmful', fontsize=13, fontweight='bold')
plt.ylabel('Churn Rate')
for i, v in enumerate(offer_stats['Churn_Rate']):
    plt.text(i, v + 0.01, f'{v:.1%}', ha='center')
plt.show()
```

```python
internet_stats = df.groupby('Internet Type', observed=True)['Churn_Flag'].mean().sort_values(ascending=False)
internet_stats = internet_stats[internet_stats.index != 'No Internet Service']

colors_internet = ['#e74c3c' if v == internet_stats.max() else '#95a5a6' for v in internet_stats.values]

plt.figure(figsize=(7, 5))
plt.bar(internet_stats.index, internet_stats.values, color=colors_internet)
plt.title('Fiber Optic: Premium Product, Highest Churn', fontsize=13, fontweight='bold')
plt.ylabel('Churn Rate')
for i, v in enumerate(internet_stats.values):
    plt.text(i, v + 0.01, f'{v:.1%}', ha='center')
plt.show()
```
